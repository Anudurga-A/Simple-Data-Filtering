Simple Data Filtering

```python
students = [
    {"name": "Anu", "age": 21, "marks": 85, "department": "MCA"},
    {"name": "Priya", "age": 22, "marks": 72, "department": "MCA"},
    {"name": "Rahul", "age": 20, "marks": 90, "department": "BCA"},
    {"name": "Kavi", "age": 23, "marks": 65, "department": "BSc"},
    {"name": "Divya", "age": 21, "marks": 88, "department": "MCA"}
]
```

```python
high_marks = []

for student in students:
    if student["marks"] > 80:
        high_marks.append(student)

print(high_marks)
```

```python
age_filter = []

for student in students:
    if student["age"] >= 22:
        age_filter.append(student)

print(age_filter)
```

```python
mca_students = []

for student in students:
    if student["department"] == "MCA":
        mca_students.append(student)

print(mca_students)
```
